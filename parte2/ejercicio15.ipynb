```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)
val manosRonda2 = Array(22, 19, 20, 21)

def bust(numero: Int): Boolean = {
  if (numero > 21) true else false
}

def clasificarMano(puntuacion: Int): String = {
  if (bust(puntuacion)) "se ha pasado"
  else if (puntuacion == 21) "21 exacto"
  else "puntuación válida"
}

def mejorPuntuacionValida(puntuaciones: Array[Int]): Int = {
  var mejor = -1
  var i = 0
  while (i < puntuaciones.length) {
    if (!bust(puntuaciones(i)) && puntuaciones(i) > mejor) {
      mejor = puntuaciones(i)
    }
    i += 1
  }
  mejor
}


def jugarRonda(titulo: String, nombres: List[String], puntuaciones: Array[Int]): Unit = {
  val mejorPuntuacion = mejorPuntuacionValida(puntuaciones)

  var i = 0
    
  while (i < puntuaciones.length) {
    val nombre = nombres(i)
    val puntos = puntuaciones(i)

    if (bust(puntos)) {
      println(s"$nombre: $puntos : Se pasa")
    } else if (puntos == mejorPuntuacion) {
      println(s"$nombre: $puntos : Mejor puntuación posible")
    } else {
      println(s"$nombre: $puntos : Puede continuar")
    }

    i += 1
  }

  if (mejorPuntuacion == -1) {
    println("No hay ganador.")
  } else {
    println(s"Mejor puntuación válida: $mejorPuntuacion")
  }
  println()
}

jugarRonda("Ronda 1", jugadores, manos)
jugarRonda("Ronda 2", jugadores, manosRonda2)
```

### Partes mutables
- Las var: mejor e i dentro de mejorPuntuacionValida, e i dentro de jugarRonda
- Los Array (manos y manosRonda2) son de un tipo mutable, porque sus elementos podrían cambiarse con manos(0) = 5

### Partes inmutables
- jugadores es una List, que no se puede modificar
- Todos los val: jugadores, manos, manosRonda2, mejorPuntuacion, nombre y puntos

### Dónde utilizo funciones
- bust: recibe una puntuación y devuelve true si supera 21
- clasificarMano: recibe una puntuación y devuelve un texto que la describe
- mejorPuntuacionValida: recibe el array de puntuaciones y devuelve la mejor que no se pasa
- jugarRonda: agrupa todas las funciones de una ronda completa y se juega

### Dónde utilizo estructuras de control
- if / else if / else: en bust, clasificarMano, mejorPuntuacionValida y jugarRonda.
- while: en mejorPuntuacionValida y en jugarRonda, para recorrer las puntuaciones con un contador.
- Operadores relacionales: >, == y <.
- Operadores lógicos: && y !, por ejemplo en !bust(puntuaciones(i)) && puntuaciones(i) > mejor.

### Parte más cercana al estilo funcional
Las funciones bust y clasificarMano. Solo dependen de su parámetro, no modifican nada externo y devuelven siempre el mismo resultado para la misma entrada (funciones puras). mejorPuntuacionValida también se comporta como función pura desde fuera, porque recibe datos y devuelve un valor sin tocar nada externo, aunque por dentro use var.

### Parte más cercana al estilo imperativo
Los bucles while con la variable contador i. También jugarRonda porque simplemente devuelve algo por pantalla
