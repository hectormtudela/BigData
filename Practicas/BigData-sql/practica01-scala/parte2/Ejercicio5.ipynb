Ejercicio 5 — Comparación de dos manos

```scala
def maxHand(numero1 : Int, numero2 : Int): Int = {
    var numeroMayor=0
    if (numero1>numero2){
        numeroMayor = numero1
    }
    else if(numero2>numero1) {
        numeroMayor = numero2
    }

    numeroMayor
}
    
```

```scala
println(maxHand(17,19))
println(maxHand(18,-1))
println(maxHand(9,9))
```

Aquí el problema es cuando los dos números son el mismo debido a que la función no modifica el valor `numeroMayor` en la función.

Para ello lo correjímos de esta manera:

def maxHand(numero1 : Int, numero2 : Int): Int = {
    var numeroMayor=0
    if (numero1>numero2){
        numeroMayor = numero1
    }
    else {
        numeroMayor = numero2
    }

    numeroMayor
}

De esta manera, si son el mismo número dará uno de los dos números que será lo que pide el programa, si el mayor es el 2º también lo dará.
