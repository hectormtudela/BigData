```scala
val jugadores = Array("Alex", "Chen", "Marta")
println(jugadores)
jugadores(0) = "Sindhu"
println(jugadores)


```

```scala
jugadores(2) = 500
```

```scala
- Por qué los elementos del array pueden cambiar aunque la variable se haya declarado con `val`: Porque lo que es inmutable es la variable jugadores en sí, no los valores de dentro.
- Por qué Scala no permite introducir el valor `500` en un `Array[String]`: Porque un Array[String] está tipado y le está metiendo un tipo Int a un String.
- La diferencia entre reasignar la variable y modificar un elemento del array: Reasignar la variable sería cambiar el array entero, modificar un elemento sería lo que hemos hecho ahora.


```
