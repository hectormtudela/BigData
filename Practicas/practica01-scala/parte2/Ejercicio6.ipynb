```scala
def bust(numero: Int): Boolean = {
    if(numero > 21) true else false
}

def ganador(handA: Int, handB: Int): Int = {

    var resultado = 0
    
    if (bust(handA) && !bust(handB))
    {
        resultado = handB
    }
    else if (!bust(handA) && bust(handB))
    {
        resultado = handA
    }
    else if (!bust(handA) && !bust(handB)){
        resultado = scala.math.max(handA, handB)
    } else {
    resultado = 0
    }

    resultado
}




println(ganador(26, 20)) // 1ª Condición
println(ganador(18, 22) // 2ª Condición)
println(ganador(24, 25 // Condición final))
println(ganador(17, 1 // 3ª Condición9))
println(ganador(21,  // 3ª Condición20))
```
