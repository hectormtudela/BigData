## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

print("Producto en cuarta fila:", inventario[3][1])
```

```python
valor_stock = []

for fila in inventario:
    producto = fila[1]
    unidades = fila[2]
    coste_unitario = fila[3]
    
    valor = round(unidades * coste_unitario, 2)
    valor_stock.append([producto, valor])

print("Valor stock:")
print(valor_stock)
```

```python
valor_inmovilizado_total = round(sum(fila[1] for fila in valor_stock), 2)

print("Valor inmovilizado total:", valor_inmovilizado_total, "€")
```

```python
inventario_sur = inventario[2:4]

valor_sur = round(
    sum(fila[2] * fila[3] for fila in inventario_sur),
    2
)

print("Inventario ALM-SUR:", inventario_sur)
print("Valor ALM-SUR:", valor_sur, "€")
```

## Ejercicio 5 — Funciones propias para el cálculo de importes

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """Calcula el importe neto de una linea de pedido."""
    bruto = unidades * precio_unitario
    importe_neto = bruto * (1 - descuento_pct / 100)
    return round(importe_neto, 2)

print("importe:", calcular_importe(3, 59.90, 10))

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """Devuelve importe bruto, ahorro e importe neto."""
    importe_bruto = unidades * precio_unitario
    ahorro = importe_bruto * descuento_pct / 100
    importe_neto = importe_bruto - ahorro
    
    return (
        round(importe_bruto, 2),
        round(ahorro, 2),
        round(importe_neto, 2)
    )

print("Resumen:", resumen_pedido(3, 59.90, 10))
```

```python
importe_bruto, ahorro, importe_neto = resumen_pedido(
    unidades=3,
    precio_unitario=59.90,
    descuento_pct=10
)

print("Importe bruto:", importe_bruto)
print("ahorro:", ahorro)
print("importe neto:", importe_neto)

help(calcular_importe)
```

```python
pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = [calcular_importe(*pedido) for pedido in pedidos]

print("Importes:", importes)
print("Importe max:", max(importes))
print("Importe min:", min(importes))
print("Importes ordenados:", sorted(importes))
print("Nº importes:", len(importes))

print("Comprobacion calcular_importe(3, 59.90, 10):", calcular_importe(3, 59.90, 10))
```

## Ejercicio 11 — Estadística descriptiva sobre telemetría simulada

```python
rng = np.random.default_rng(2024)

temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

print("Nº de temperaturas:", len(temperatura))
print("Nº de mediciones de vibraciom:", len(vibracion))
```

```python
media_temp = np.mean(temperatura)
mediana_temp = np.median(temperatura)
desviacion_temp = np.std(temperatura)
min_temp = np.min(temperatura)
max_temp = np.max(temperatura)

print("Media:", round(media_temp, 2))
print("Mediana:", round(mediana_temp, 2))
print("Desviación estándar:", round(desviacion_temp, 2))
print("Mínimo:", round(min_temp, 2))
print("Máximo:", round(max_temp, 2))

print("La media y la mediana son similares, lo que indica una distribución sin muchas diferencias entre ambas medidas.")
```

```python
percentiles_vibracion = np.percentile(vibracion, [25, 50, 75, 95])

print("Percentil 25:", round(percentiles_vibracion[0], 2))
print("Percentil 50:", round(percentiles_vibracion[1], 2))
print("Percentil 75:", round(percentiles_vibracion[2], 2))
print("Percentil 95:", round(percentiles_vibracion[3], 2))

telemetria = np.column_stack((temperatura, vibracion))

print("Forma matriz telemetría:", telemetria.shape)
print("Primeras 5 filas:")
print(telemetria[:5])
```

```python
correlacion = np.corrcoef(temperatura, vibracion)[0, 1]

print("Correlacion entre temperatura y vibración:", round(correlacion, 3))
print("La correlacion mide la relación lineal entre ambas variables.")

umbral_temperatura = media_temp + 2 * desviacion_temp
temperaturas_altas = temperatura > umbral_temperatura

numero_altas = np.sum(temperaturas_altas)
porcentaje_altas = numero_altas / len(temperatura) * 100

print("Umbral temperatura:", round(umbral_temperatura, 2))
print("Nº temperaturas > que umbral:", numero_altas)
print("Porcentaje:", round(porcentaje_altas, 2), "%")
```

## Ejercicio 15 — Columnas calculadas, agregación y exportación

```python
ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]

ventas["importe_neto"] = round(
    ventas["importe_bruto"] * (1 - ventas["descuento_pct"] / 100),
    2
)

ventas["ticket_medio"] = round(
    ventas["importe_neto"] / ventas["unidades"],
    2
)

print(ventas[["importe_bruto", "importe_neto", "ticket_medio"]].head())
```

```python
facturacion_region = (
    ventas.groupby("region")["importe_neto"]
    .sum()
    .sort_values(ascending=False)
)

print("facturacion por region:")
print(facturacion_region)
```

```python
resumen_categoria_canal = (
    ventas.groupby(["categoria", "canal"])
    .agg(
        importe_neto_medio=("importe_neto", "mean"),
        unidades_totales=("unidades", "sum")
    )
)

print(resumen_categoria_canal)
```

```python
top_5_productos = (
    ventas.groupby("producto")["importe_neto"]
    .sum()
    .nlargest(5)
)

print("Top 5 productos x facturacion neta:")
print(top_5_productos)
```

```python
ventas["satisfaccion"] = ventas["satisfaccion"].fillna(
    ventas["satisfaccion"].median()
)

ventas["canal"] = ventas["canal"].fillna("Desconocido")

print("valores nulos de satisfaccion:", ventas["satisfaccion"].isna().sum())
print("valores nulos de canal:", ventas["canal"].isna().sum())
```

```python
temp = np.array([68, 72, 75, 81, 69])
vib = np.array([2.1, 3.2, 2.8, 3.7, 1.9])

try:
    resultado = temp > 70 and vib > 3
    print(resultado)
except ValueError as e:
    print("Error:", e)
    print("Con arrays de numpy se debe usar una operacion elemento a elemento.")
```

```python
resultado_logical = np.logical_and(temp > 70, vib > 3)
resultado_operador = (temp > 70) & (vib > 3)

print("Con np.logical_and():", resultado_logical)
print("Con &: ", resultado_operador)
print("Son iguales?:", np.array_equal(resultado_logical, resultado_operador))
```

## Ejercicio 17 — `if` / `elif` / `else` aplicado a mantenimiento predictivo

```python
def nivel_riesgo(temperatura):
    if temperatura > 85:
        return "CRITICO"
    elif temperatura > 76:
        return "ALTO"
    elif temperatura > 70:
        return "MEDIO"
    else:
        return "BAJO"

temperaturas_prueba = [92.5, 80, 72.3, 64.1]

for temperatura in temperaturas_prueba:
    print(temperatura, "°C ==", nivel_riesgo(temperatura))
```

```python
sensores["nivel_riesgo"] = sensores["temperatura_c"].apply(nivel_riesgo)

print(sensores[["temperatura_c", "nivel_riesgo"]].head())

print("\nNúmero de sensores por nivel:")
print(sensores["nivel_riesgo"].value_counts())
```

Ante niveles de riesgo altos o críticos, se recomienda priorizar la revision de los equipos afectados. El seguimiento de la temperatura permite detectar situaciones que requieren mantenimiento antes de que aumente el riesgo.
