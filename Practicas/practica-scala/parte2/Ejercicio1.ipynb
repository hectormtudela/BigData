EJERCICIO 01

Crea variables que representen la información de una partida de cartas:

- Nombre del jugador.
- Número de partidas jugadas.
- Puntuación actual.
- Porcentaje de victorias.
- Indicación de si el jugador continúa activo.

VERSION A

```scala
val jugador: String = "Héctor"
val nPartidasJugadas: Int = 0
val puntuacionActual: Int = 0
val porcentajeVictorias: Float = 0
val jugadorActivo: Boolean = true

println(s"Nombre: $jugador | Partidas Jugadas: $nPartidasJugadas | Porcentaje de victorias: $porcentajeVictorias")
```

VERSION B

```scala
var jugador = "Héctor"
var nPartidasJugadas = 0
var puntuacionActual = 0
var porcentajeVictorias = 0.0
var jugadorActivo = true

println(s"Nombre: $jugador | Partidas Jugadas: $nPartidasJugadas | Porcentaje de victorias: $porcentajeVictorias")
```
