Ejercicio 9 — Recorrer un Array con while

```scala
val manos = Array(17, 24, 21, 19, 26)

def bust(numero: Int): Boolean = {
    if(numero > 21) true else false
}

var i = 0

while (i < manos.length) {
    println(manos(i))
    println(bust(manos(i)))
    i = i + 1
    
}

```

- Por qué el contador debe ser mutable. -> Para poder sumarle 1 cada iteracion para mostrar todos los datos de un array.
- Qué ocurriría si no incrementaras `i`. -> Se haría un bucle infinito en el que mostraría solo el primer valor del array.
- Qué elementos del ejercicio corresponden al estilo imperativo. -> Todo el while es un elemento imperativo. Las variables `var` tambien lo son
