Ejercicio 2 — val, var y reasignación


Dado el código, realiza las siguientes operaciones:
1. Incrementa la puntuación en 5 puntos.
2. Vuelve a incrementarla en 3 puntos.
3. Muestra el resultado final.
4. Intenta reasignar el valor de `jugador`.

La última operación deberá producir un error.

```scala
val jugador = "Marta"
var puntuacion = 10

puntuacion += 5
println(puntuacion)

puntuacion += 3
println(puntuacion)

```

```scala
jugador = "Pepe"
```

Puntuación puede modificarse debido a que es un `var`.

Jugador no puede modificarse debido a que es un `val`.

Scala genera este código de error: 

cmd4.sc:1: reassignment to val
val res4 = jugador = "Pepe"
                   ^
Compilation Failed
