```scala

var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

println(s"Total inicial: $total")

sumarAlTotal(5)
println(total)

sumarAlTotal(10)
println(total)

sumarAlTotal(3)
println(total)
```

```scala

def sumar(a: Int, b: Int): Int = {
  a + b
}

val r1 = sumar(0, 5)
println(r1)
val r2 = sumar(r1, 10)
println(r2)
val r3 = sumar(r2, 3)
println(r3)

println(s"El total externo sigue siendo 18")  
```

| Característica | `sumarAlTotal` | `sumar` |
| --- | --- | --- |
| Modifica datos externos | si | no |
| Devuelve un resultado calculado | no | si |
| Utiliza efecto secundario | si | no |
| Estilo predominante | imperativo | funcional |
